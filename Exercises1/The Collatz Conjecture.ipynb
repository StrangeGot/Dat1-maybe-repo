I define 2 lists; one hosting all the numbers in the process, and one for the amount of steps taken

```python

allnum = []
steps = []
```

I define a function that corresponds to the rules

```python
for i in range(1,101):
    n=i
    s=0
    while i>1:
        #print(i)
        allnum.append([n,i])
        if i%2 == 0:
            i=i*1/2
        else:
            i=i*3+1
        s+=1
    steps.append([n,s])
    print(f"{n} becomes {i}")
```

Now, i find i check the lists for the highest number, and the most steps taken, while also knowing which number it corresponds to. I do this by first defining a function that find the row, with the highest value in the second column.

```python
def correspond(row):
    return row[1]
moststeps = max(steps, key=correspond)
highestnum = max(allnum, key=correspond)
print(f"Highest integer achieved it {highestnum[1]}, and it's from {highestnum[0]}")
print(f"Most steps used is {moststeps[1]}, and it's from {moststeps[0]}")
```
